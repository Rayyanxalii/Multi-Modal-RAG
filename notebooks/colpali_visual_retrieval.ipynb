```python
from pathlib import Path
from PIL import Image
import torch

image_dir = Path("/content/rendered_pages/rendered_pages")

image_paths = sorted(
    p for p in image_dir.rglob("*")
    if p.suffix.lower() in {".png", ".jpg", ".jpeg"}
)

assert image_paths, f"No images found in {image_dir}"

# Start with three pages to limit GPU memory usage.
images = [
    Image.open(path).convert("RGB")
    for path in image_paths[:3]
]

with torch.inference_mode():
    page_embeddings = model.encode_document(
        images,
        batch_size=1,
        convert_to_tensor=True,
        show_progress_bar=True,
    )

print("Pages encoded:", len(page_embeddings))
print("First page embedding shape:", page_embeddings[0].shape)
```

```python
from sentence_transformers import MultiVectorEncoder
import torch

model = MultiVectorEncoder("vidore/colpali-v1.3-hf")

print("Model class:", type(model))
print("Supported modalities:", model.modalities)
print("Image support:", model.supports("image"))
print("CUDA available:", torch.cuda.is_available())
```

```python
from PIL import Image
import torch

images = [
    Image.open(path).convert("RGB")
    for path in image_paths
]

with torch.inference_mode():
    page_embeddings = model.encode_document(
        images,
        batch_size=2,
        show_progress_bar=True,
    )

print("Total pages encoded:", len(page_embeddings))
print("First page embedding shape:", page_embeddings[0].shape)
print("Last page embedding shape:", page_embeddings[-1].shape)
```

```python
query = ["What is the attention mechanism?"]

with torch.inference_mode():
    query_embeddings = model.encode_query(
        query,
        show_progress_bar=True,
    )

    scores = model.similarity(query_embeddings, page_embeddings)

print("Query embedding shape:", query_embeddings[0].shape)
print("\nPage relevance scores:")

ranked_indices = scores[0].argsort(descending=True).tolist()

for rank, idx in enumerate(ranked_indices, start=1):
    print(
        f"Rank {rank}: {image_paths[idx].name} "
        f"| Score: {scores[0][idx].item():.4f}"
    )
```

```python
from IPython.display import display

for idx in ranked_indices:
    print(
        f"Ranked page: {image_paths[idx].name} "
        f"| Score: {scores[0][idx].item():.4f}"
    )
    display(images[idx])
```
